```python
import numpy as np
import scipy.optimize as opt
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python
# 1. Veri Yükleme
traffic_data = np.array([12, 15, 10, 8, 14, 11, 13, 16, 9, 12, 11, 14, 10, 15])

# 2. Negatif Log-Likelihood Fonksiyonu
def negative_log_likelihood(lam, data):
    n = len(data)
    # log(k!) sabit olduğu için ihmal edilmiştir
    # Formül: -(-n*lam + log(lam)*sum(data))
    nll = -(-n * lam + np.log(lam) * np.sum(data))
    return nll

# 3. Optimizasyon
initial_guess = 1.0
result = opt.minimize(negative_log_likelihood, initial_guess, args=(traffic_data,), bounds=[(0.001, None)])
mle_lambda = result.x[0]

print(f"Sayısal Tahmin (MLE lambda): {mle_lambda}")
print(f"Analitik Tahmin (Ortalama): {np.mean(traffic_data)}")
```

```python
# Bölüm 3: Görselleştirme
plt.figure(figsize=(10, 6))
# Gerçek veri histogramı
plt.hist(traffic_data, bins=range(min(traffic_data), max(traffic_data) + 2),
         density=True, alpha=0.6, color='skyblue', label='Gerçek Veri (Histogram)', align='left')

# Poisson PMF grafiği
x = np.arange(0, 25)
plt.plot(x, poisson.pmf(x, mle_lambda), 'ro--', label=f'Poisson PMF ($\lambda$={mle_lambda:.2f})')

plt.xlabel('Bir Dakikadaki Araç Sayısı')
plt.ylabel('Olasılık')
plt.title('Trafik Yoğunluğu: Gerçek Veri vs Poisson Modeli')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
